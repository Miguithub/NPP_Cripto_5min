## Diccionario de `df`
Ejemplo para BTC; el mismo esquema se aplica a cada activo base observado ese mes.

| Columna | Unidad / significado |
|---|---|
| `timestamp` (índice) | UTC; momento de disponibilidad asumida con margen |
| `bar_start`, `bar_end` | Inicio y final real de la vela |
| `BTC_y` | Cierre en USDT × volumen en BTC; columna original de `dftu` |
| `YS` | Suma observada de las columnas `_y` |
| `BTCUSDT` | USDT por BTC; directo o sintético según bandera |
| `BTC_v` | BTC negociados en el mercado seleccionado |
| `BTC_yp` | Participación entre 0 y 1 sobre `YS` |
| `BTC_source_pair` | Par que aporta precio y volumen; nulo si no hay dato utilizable |
| `BTC_price_is_proxy` | Verdadero si no se usa directamente BTCUSDT |
| `BTC_observed` | Hay vela y precio utilizable en el mercado seleccionado |
| `BTC_active_trading` | Número de operaciones > 0 en ese mercado; nulo sin observación |
| `BTC_turnover_usdt_proxy` | `quote_volume × quote/USDT`; proxy de liquidez por actividad |
| `BTC_range_rel_proxy` | `(high−low)/close`, adimensional; rango, no spread |
| `BTC_illiquidity_proxy` | `abs(close/open−1)/turnover_usdt`; variante intravela, inverso de USDT |
| `BTC_activity_trades_proxy` | Cantidad de operaciones; NO personas ni cuentas únicas |
| `BTC_taker_buy_usdt_proxy` | Flujo de compras agresoras en USDT aproximados |
| `BTC_taker_sell_usdt_proxy` | Flujo de ventas agresoras en USDT aproximados |
| `BTC_active_users` | `NaN`: dato no disponible |
| `BTC_spot_account_liquidity` | `NaN`: saldo agregado de cuentas spot no disponible |
| `BTC_exchange_reserves` | `NaN`: reservas históricas 5m no disponibles |
| `BTC_orderbook_depth` | `NaN`: no se reconstruye profundidad con velas |

Los flujos de takers no son depósitos/retiros ni cambios del saldo custodiado. Los proxies
de actividad y liquidez no permiten inferir los stocks de cuentas del paso 6.

#Variables de Segundo Orden, 1ra generacion

```python
# CELDA 1   → sistema
# CELDA 2   → precio + volumen
# CELDA 3   → actividad
# CELDA 4   → buy/sell
# CELDA 5   → absorción + liquidez
# CELDA 6   → impacto + stress
# CELDA 7   → gp
# CELDA 8   → gp multiescala
# CELDA 9   → N temporal
# CELDA 10  → N competitivo/anidado
# CELDA 11  → N_context + N_theory + E + target
# CELDA 12  → dinámica E_context
# CELDA 13  → dinámica E_theory
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    # 5 minutos por barra
    primary_window: int = 288      # 24 h
    min_periods: int = 72          # 6 h
    ewm_span: int = 72             # 6 h

    multiscale_windows: tuple[int, ...] = (
        3,      # 15 min
        12,     # 1 h
        72,     # 6 h
        288,    # 24 h
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    # Reduce RAM al trabajar con 510 activos
    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):
    """
    Añade todas las features de una celda en un único concat.
    Permite volver a ejecutar una celda sin duplicar columnas.
    """

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):
    """
    Compara x_t exclusivamente contra historia t-1,...,t-W.
    """

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):
    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    # Impacto negativo no tiene interpretación
    # como "más resistencia".
    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):

    g = gp0.to_numpy(
        dtype=np.float64,
        copy=False
    )

    s = stress.to_numpy(
        dtype=np.float64,
        copy=False
    )

    q = stress_threshold.to_numpy(
        dtype=np.float64,
        copy=False
    )

    b = baseline.to_numpy(
        dtype=np.float64,
        copy=False
    )

    T, N = g.shape

    output = np.full(
        (T, N),
        np.nan
    )

    for j in range(N):

        in_event = False

        event_start = -1
        event_baseline = np.nan

        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):

            if not in_event:

                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):

                    in_event = True

                    event_start = t
                    event_baseline = b[t, j]

                    consecutive = 0

            else:

                if np.isfinite(g[t, j]):

                    distance = (
                        abs(
                            g[t, j]
                            - event_baseline
                        )
                        /
                        (
                            abs(event_baseline)
                            + eps
                        )
                    )

                    if distance <= tolerance:
                        consecutive += 1
                    else:
                        consecutive = 0

                    if consecutive >= hold_bars:

                        recovery_start = (
                            t
                            - hold_bars
                            + 1
                        )

                        duration = max(
                            1,
                            recovery_start
                            - event_start
                        )

                        if np.isnan(duration_ewma):

                            duration_ewma = float(
                                duration
                            )

                        else:

                            duration_ewma = (
                                alpha * duration
                                + (1-alpha)
                                * duration_ewma
                            )

                        in_event = False

                        event_start = -1
                        event_baseline = np.nan
                        consecutive = 0

            if np.isfinite(duration_ewma):

                output[t, j] = (
                    1.0
                    / (1.0 + duration_ewma)
                )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns
    )


# ============================================================
# INICIALIZACIÓN
# ============================================================

df = validate_time(df, cfg)

assets = discover_assets(
    df,
    cfg
)

print("Activos detectados:", len(assets))
```

```python
blocks = []


for chunk in chunks(
    assets,
    cfg.asset_block_size
):

    O = load_base(
        df, chunk, "O", cfg
    )

    A = load_base(
        df, chunk, "A", cfg
    )

    observed = O > 0

    P = load_base(
        df, chunk, "P", cfg
    ).where(observed)

    V = load_base(
        df, chunk, "V", cfg
    ).where(observed)


    # --------------------------------------------
    # Retorno
    # --------------------------------------------

    logP = np.log(
        P.where(P > 0)
    )

    ret = logP.diff(
        cfg.return_lag
    )


    # --------------------------------------------
    # Volatilidad rolling
    # --------------------------------------------

    volatility = (
        ret
        .rolling(
            cfg.primary_window,
            min_periods=cfg.min_periods
        )
        .std()
    )


    # --------------------------------------------
    # Crecimiento de volumen físico
    # --------------------------------------------

    V_valid = V.where(
        (V > 0)
        & (A > 0)
    )

    dlogV = (
        np.log(V_valid)
        .diff()
    )


    # --------------------------------------------
    # Estabilidad del volumen
    # --------------------------------------------

    volume_dispersion = (
        rolling_robust_dispersion(
            dlogV,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )
    )

    VSTAB = np.exp(
        -volume_dispersion
    )


    blocks.extend([
        name_block(
            ret,
            "return_k1",
            cfg
        ),

        name_block(
            volatility,
            "rolling_volatility",
            cfg
        ),

        name_block(
            dlogV,
            "dlogV",
            cfg
        ),

        name_block(
            VSTAB,
            "VSTAB",
            cfg
        ),
    ])


df = merge_blocks(
    df,
    blocks
)
```

```python
blocks = []


for chunk in chunks(
    assets,
    cfg.asset_block_size
):

    O = load_base(
        df, chunk, "O", cfg
    )

    observed = O > 0

    B = load_base(
        df, chunk, "B", cfg
    ).where(observed)

    S = load_base(
        df, chunk, "S", cfg
    ).where(observed)


    # --------------------------------------------
    # Sell intensity
    # --------------------------------------------

    logS1 = np.log1p(
        S.clip(lower=0)
    )

    SI = robust_z_causal(
        logS1,
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )


    # --------------------------------------------
    # Sell growth
    # --------------------------------------------

    SG = (
        np.log(
            S.clip(lower=0)
            + cfg.eps
        )
        .diff()
    )


    # --------------------------------------------
    # Sell persistence
    # --------------------------------------------

    SP = (
        SI
        .ewm(
            span=cfg.ewm_span,
            adjust=False,
            min_periods=cfg.min_periods
        )
        .mean()
    )


    # --------------------------------------------
    # Buy/Sell
    # --------------------------------------------

    total = B + S

    imbalance = (
        (B - S)
        /
        (total + cfg.eps)
    )

    sell_share = (
        S
        /
        (total + cfg.eps)
    )

    LR_BS = np.log(
        (B + cfg.eps)
        /
        (S + cfg.eps)
    )

    BIL = (
        2.0
        * np.minimum(B, S)
        /
        (total + cfg.eps)
    )


    blocks.extend([
        name_block(SI, "SI", cfg),
        name_block(SG, "SG", cfg),
        name_block(SP, "SP", cfg),
        name_block(imbalance, "imbalance", cfg),
        name_block(sell_share, "sell_share", cfg),
        name_block(LR_BS, "LR_BS", cfg),
        name_block(BIL, "BIL", cfg),
    ])


df = merge_blocks(
    df,
    blocks
)
```

##5. Bilaterailidad, absorcion y liquidez

- Estabilidad Bilateralidad

BILST = EWMA(BIL)*exp(MADw(delta(Bil)))

Persistencia de un mercado bilateral


ABS
ABS_S
RLIQ
RLIQ_S
LIQ

```python
blocks = []


for chunk in chunks(
    assets,
    cfg.asset_block_size
):

    O = load_base(
        df, chunk, "O", cfg
    )

    observed = O > 0

    C = load_base(
        df, chunk, "C", cfg
    ).where(observed)

    S = load_base(
        df, chunk, "S", cfg
    ).where(observed)

    R = load_base(
        df, chunk, "R", cfg
    ).where(observed)

    L = load_base(
        df, chunk, "L", cfg
    ).where(observed)

    BIL = load_feature(
        df, chunk, "BIL"
    )


    # --------------------------------------------
    # Bilaterality stability
    # --------------------------------------------

    BIL_disp = rolling_robust_dispersion(
        BIL.diff(),
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    BILST = (
        BIL
        .ewm(
            span=cfg.ewm_span,
            adjust=False,
            min_periods=cfg.min_periods
        )
        .mean()
        *
        np.exp(-BIL_disp)
    )


    logC = np.log1p(
        C.clip(lower=0)
    )

    logS = np.log1p(
        S.clip(lower=0)
    )

    logL = np.log1p(
        L.clip(lower=0)
    )


    ABS = (
        logC
        /
        (R + cfg.eps)
    )

    ABS_S = (
        logS
        /
        (R + cfg.eps)
    )

    RLIQ = (
        logC
        /
        (1.0 + logL)
    )

    RLIQ_S = (
        logS
        /
        (1.0 + logL)
    )

    LIQ = -logL


    blocks.extend([
        name_block(BILST, "BILST", cfg),
        name_block(ABS, "ABS", cfg),
        name_block(ABS_S, "ABS_S", cfg),
        name_block(RLIQ, "RLIQ", cfg),
        name_block(RLIQ_S, "RLIQ_S", cfg),
        name_block(LIQ, "LIQ", cfg),
    ])


df = merge_blocks(
    df,
    blocks
)
```

##6. Impacto, stress y supervivencia
- Impact resistence

IR

- Shock Externo:

STRESS = Z_W(R)+Z_W(L)+Z_W(r)

Medir pertubacion, no resistencia

- Survival under stress:

SSR = Median(C|STRESS>q)/ Median(C) +epsilon

cuanta actividad conserva bajo stress

- Ventas sobrevientes bajo schock:

SSR_S = Median(S|STRESS>q)/ Median(S) +epsilon

Cuánta presión vendedora puede absorber en stress

- Venta media por transaccion

AST_S = S/ C + epsilon

Tamaño monetario vendedor por transacción observada


a exluir:
""
FLOW_TOTAL_AUDIT: B+S
TURNOVER_PER_TRADE_AUDIT
""

```python
blocks = []


for chunk in chunks(
    assets,
    cfg.asset_block_size
):

    O = load_base(
        df, chunk, "O", cfg
    )

    observed = O > 0

    C = load_base(
        df, chunk, "C", cfg
    ).where(observed)

    R = load_base(
        df, chunk, "R", cfg
    ).where(observed)

    L = load_base(
        df, chunk, "L", cfg
    ).where(observed)

    B = load_base(
        df, chunk, "B", cfg
    ).where(observed)

    S = load_base(
        df, chunk, "S", cfg
    ).where(observed)


    ret = load_feature(
        df, chunk, "return_k1"
    )

    imbalance = load_feature(
        df, chunk, "imbalance"
    )


    # --------------------------------------------
    # Impact resistance
    # --------------------------------------------

    logC = np.log1p(
        C.clip(lower=0)
    )

    pressure = (
        imbalance.abs()
        * logC
    )

    beta = rolling_beta_with_intercept(
        pressure,
        ret.abs(),
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    IR = (
        1.0
        /
        (beta + cfg.eps)
    )


    # --------------------------------------------
    # Stress
    # --------------------------------------------

    zR = robust_z_causal(
        R,
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    zL = robust_z_causal(
        L,
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    zRet = robust_z_causal(
        ret.abs(),
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    STRESS = (
        zR
        + zL
        + zRet
    )


    threshold = rolling_stress_threshold(
        STRESS,
        cfg.primary_window,
        cfg.min_periods,
        cfg.stress_quantile
    )


    # --------------------------------------------
    # Survival
    # --------------------------------------------

    SSR = conditional_rolling_median_ratio(
        C,
        STRESS,
        threshold,
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )

    SSR_S = conditional_rolling_median_ratio(
        S,
        STRESS,
        threshold,
        cfg.primary_window,
        cfg.min_periods,
        cfg.eps
    )


    # --------------------------------------------
    # Average sell per trade
    # --------------------------------------------

    AST_S = (
        S
        /
        (C + cfg.eps)
    )


    # --------------------------------------------
    # Auditoría
    # --------------------------------------------

    # FLOW_TOTAL = B + S

    # TURNOVER_PER_TRADE = (
    #     FLOW_TOTAL
    #     /
    #     (C + cfg.eps)
    # )


    blocks.extend([
        name_block(IR, "IR", cfg),
        name_block(STRESS, "STRESS", cfg),
        name_block(SSR, "SSR", cfg),
        name_block(SSR_S, "SSR_S", cfg),
        name_block(AST_S, "AST_S", cfg),

        # name_block(
        #     FLOW_TOTAL,
        #     "FLOW_TOTAL_AUDIT",
        #     cfg
        # ),

        # name_block(
        #     TURNOVER_PER_TRADE,
        #     "TURNOVER_PER_TRADE_AUDIT",
        #     cfg
        # ),
    ])


df = merge_blocks(
    df,
    blocks
)
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c1


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c3


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from pathlib import Path
import os
import gc


# ============================================================
# DATASETS DE TRABAJO
# ============================================================

DATASETS = {
    "df_c1": df_c1,
    "df_c2": df_c2,
    "df_c3": df_c3,
}


# ============================================================
# ARCHIVOS DE SALIDA
# ============================================================

ruta = Path(
    "/content/drive/MyDrive/Neural/NPP/Cripto/Crypto 5 minutos/"
)

OUTPUTS = {
    "df_c1": ruta / "112024-042025_clean_add_gp_fgp_nmetrics.parquet",
    "df_c2": ruta / "042025-092025_clean_add_gp_fgp_nmetrics.parquet",
    "df_c3": ruta / "092025-022026_clean_add_gp_fgp_nmetrics.parquet",
}


# ============================================================
# CONTROL DE PROCESO
# ============================================================

saved_files = []
completed_datasets = []

current_dataset = None
current_chunk = 0
total_chunks = 0


try:

    for dataset_name, DATASET in DATASETS.items():

        current_dataset = dataset_name

        asset_chunks = list(
            chunks(
                assets,
                cfg.asset_block_size
            )
        )

        total_chunks = len(asset_chunks)
        current_chunk = 0

        print("\n" + "=" * 70)
        print(
            f"{dataset_name} | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas | "
            f"{total_chunks} chunks"
        )
        print("=" * 70)


        for i, chunk in enumerate(
            asset_chunks,
            start=1
        ):

            # ====================================================
            # CÁLCULO DEL CHUNK
            # ====================================================

            gp = load_feature(
                DATASET,
                chunk,
                "gp"
            )


            # --------------------------------------------
            # Estabilidad gp
            # --------------------------------------------

            dlog_gp = (
                np.log(
                    gp.clip(
                        lower=cfg.eps
                    )
                )
                .diff()
            )

            gp_dispersion = (
                rolling_robust_dispersion(
                    dlog_gp,
                    cfg.primary_window,
                    cfg.min_periods,
                    cfg.eps
                )
            )

            NSTAB = np.exp(
                -gp_dispersion
            )


            # --------------------------------------------
            # Persistencia gp
            # --------------------------------------------

            NPERS = rolling_autocorr_lag1(
                gp,
                cfg.primary_window,
                cfg.min_periods
            )


            # --------------------------------------------
            # Coherencia multiescala
            # --------------------------------------------

            logs = []

            for w in cfg.multiscale_windows:

                G = load_feature(
                    DATASET,
                    chunk,
                    f"gp_W{w}"
                )

                logs.append(
                    np.log(
                        G.clip(
                            lower=cfg.eps
                        )
                    )
                )


            stack = np.stack(
                [
                    x.to_numpy()
                    for x in logs
                ],
                axis=2
            )

            variance = np.nanvar(
                stack,
                axis=2
            )

            NCOH = pd.DataFrame(
                np.exp(-variance),
                index=DATASET.index,
                columns=chunk
            )


            # --------------------------------------------
            # Retención bajo stress
            # --------------------------------------------

            STRESS = load_feature(
                DATASET,
                chunk,
                "STRESS"
            )

            threshold = rolling_stress_threshold(
                STRESS,
                cfg.primary_window,
                cfg.min_periods,
                cfg.stress_quantile
            )

            NRET = conditional_rolling_median_ratio(
                gp,
                STRESS,
                threshold,
                cfg.primary_window,
                cfg.min_periods,
                cfg.eps
            )


            # ====================================================
            # PREPARACIÓN DEL CHUNK
            # ====================================================

            chunk_blocks = [
                name_block(NSTAB, "NSTAB", cfg),
                name_block(NPERS, "NPERS", cfg),
                name_block(NCOH, "NCOH", cfg),
                name_block(NRET, "NRET", cfg),
            ]


            # ====================================================
            # COMMIT DEL CHUNK
            # Solo se ejecuta si TODO el chunk terminó correctamente
            # ====================================================

            DATASET = merge_blocks(
                DATASET,
                chunk_blocks
            )

            DATASETS[dataset_name] = DATASET
            globals()[dataset_name] = DATASET

            current_chunk = i

            print(
                f"{dataset_name} | "
                f"chunk {i:02d}/{total_chunks} OK | "
                f"{len(chunk):02d} activos | "
                f"{DATASET.shape[1]:,} columnas"
            )


            del (
                gp,
                dlog_gp,
                gp_dispersion,
                NSTAB,
                NPERS,
                logs,
                G,
                stack,
                variance,
                NCOH,
                STRESS,
                threshold,
                NRET,
                chunk_blocks,
            )

            gc.collect()


        # ========================================================
        # GUARDADO DEL DATASET COMPLETO
        # ========================================================

        output_file = OUTPUTS[dataset_name]
        temp_file = output_file.with_suffix(".tmp.parquet")

        DATASET.to_parquet(
            temp_file,
            index=False
        )

        os.replace(
            temp_file,
            output_file
        )

        saved_files.append(str(output_file))
        completed_datasets.append(dataset_name)

        print(
            f"\n{dataset_name} COMPLETO | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas"
        )

        print(
            f"Guardado: {output_file}"
        )


except KeyboardInterrupt:

    print("\nPROCESO INTERRUMPIDO MANUALMENTE")


except Exception as e:

    print(
        f"\nERROR: {type(e).__name__}: {e}"
    )


finally:

    # ============================================================
    # ESTADO FINAL
    # ============================================================

    print("\n" + "=" * 70)
    print("ESTADO FINAL")
    print("=" * 70)

    if current_dataset is not None:

        print(
            f"Último proceso: {current_dataset} | "
            f"chunks confirmados: "
            f"{current_chunk}/{total_chunks}"
        )


    if completed_datasets:

        print(
            "\nDatasets completados:",
            ", ".join(completed_datasets)
        )

    else:

        print(
            "\nDatasets completados: ninguno"
        )


    if saved_files:

        print(
            "\nArchivos guardados exitosamente:"
        )

        for file in saved_files:

            print(
                f"  ✓ {file}"
            )

    else:

        print(
            "\nArchivos completos guardados: ninguno"
        )
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c1


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c3


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from pathlib import Path
import os
import gc


# ============================================================
# DATASETS DE TRABAJO
# ============================================================

DATASETS = {
    "df_c1": df_c1,
    "df_c2": df_c2,
    "df_c3": df_c3,
}


# ============================================================
# ARCHIVOS DE SALIDA
# ============================================================

ruta = Path(
    "/content/drive/MyDrive/Neural/NPP/Cripto/Crypto 5 minutos/"
)

OUTPUTS = {
    "df_c1": ruta / "112024-042025_clean_add_final_Ectx.parquet",
    "df_c2": ruta / "042025-092025_clean_add_final_Ectx.parquet",
    "df_c3": ruta / "092025-022026_clean_add_final_Ectx.parquet",
}


# ============================================================
# CONTROL DE PROCESO
# ============================================================

saved_files = []
completed_datasets = []

current_dataset = None
current_stage = None
current_chunk = 0
total_chunks = 0


try:

    for dataset_name, DATASET in DATASETS.items():

        current_dataset = dataset_name
        current_stage = "E_context features"

        asset_chunks = list(
            chunks(
                assets,
                cfg.asset_block_size
            )
        )

        total_chunks = len(asset_chunks)
        current_chunk = 0

        print("\n" + "=" * 70)
        print(
            f"{dataset_name} | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas | "
            f"{total_chunks} chunks"
        )
        print("=" * 70)


        # ============================================================
        # E_context GLOBAL
        # ============================================================

        E_all = load_feature(
            DATASET,
            assets,
            "E_context"
        ).astype("float32")


        A_all = load_base(
            DATASET,
            assets,
            "A",
            cfg
        ).astype("float32")


        E_active = E_all.where(
            A_all > 0
        )


        # --------------------------------------------
        # Estadísticos transversales
        # --------------------------------------------

        E_median = E_active.median(
            axis=1
        )

        E_rank = E_active.rank(
            axis=1,
            pct=True,
            method="average"
        )


        # ============================================================
        # CÁLCULO POR CHUNK
        # ============================================================

        for i, chunk in enumerate(
            asset_chunks,
            start=1
        ):

            E = E_all.loc[
                :,
                chunk
            ]


            # --------------------------------------------
            # Log energía
            # --------------------------------------------

            logE = np.log(
                E.clip(
                    lower=cfg.eps
                )
            )


            # --------------------------------------------
            # Energía relativa transversal
            # --------------------------------------------

            Erel = E.sub(
                E_median,
                axis=0
            )


            # --------------------------------------------
            # Primera diferencia
            # --------------------------------------------

            dE = E.diff()


            # --------------------------------------------
            # Segunda diferencia
            # --------------------------------------------

            ddE = dE.diff()


            # --------------------------------------------
            # Shock robusto causal
            # --------------------------------------------

            ESH = robust_z_causal(
                E,
                cfg.primary_window,
                cfg.min_periods,
                cfg.eps
            )


            # --------------------------------------------
            # Ranking transversal
            # --------------------------------------------

            ERANK = E_rank.loc[
                :,
                chunk
            ]


            # ========================================================
            # PREPARACIÓN DEL CHUNK
            # ========================================================

            chunk_blocks = [
                name_block(
                    logE,
                    "Ectx_logE",
                    cfg
                ),

                name_block(
                    Erel,
                    "Ectx_Erel",
                    cfg
                ),

                name_block(
                    dE,
                    "Ectx_dE",
                    cfg
                ),

                name_block(
                    ddE,
                    "Ectx_ddE",
                    cfg
                ),

                name_block(
                    ESH,
                    "Ectx_ESH",
                    cfg
                ),

                name_block(
                    ERANK,
                    "Ectx_ERANK",
                    cfg
                ),
            ]


            # ========================================================
            # COMMIT DEL CHUNK
            # Solo se ejecuta si TODO el chunk terminó correctamente
            # ========================================================

            DATASET = merge_blocks(
                DATASET,
                chunk_blocks
            )

            DATASETS[dataset_name] = DATASET
            globals()[dataset_name] = DATASET

            current_chunk = i

            print(
                f"{dataset_name} | "
                f"chunk {i:02d}/{total_chunks} OK | "
                f"{len(chunk):02d} activos | "
                f"{DATASET.shape[1]:,} columnas"
            )


            del (
                E,
                logE,
                Erel,
                dE,
                ddE,
                ESH,
                ERANK,
                chunk_blocks,
            )

            gc.collect()


        # ============================================================
        # LIMPIEZA DE OBJETOS GLOBALES DEL DATASET
        # ============================================================

        del (
            E_all,
            A_all,
            E_active,
            E_median,
            E_rank,
        )

        gc.collect()


        # ============================================================
        # GUARDADO DEL DATASET COMPLETO
        # ============================================================

        output_file = OUTPUTS[
            dataset_name
        ]

        temp_file = output_file.with_suffix(
            ".tmp.parquet"
        )

        DATASET.to_parquet(
            temp_file,
            index=False
        )

        os.replace(
            temp_file,
            output_file
        )

        saved_files.append(
            str(output_file)
        )

        completed_datasets.append(
            dataset_name
        )

        print(
            f"\n{dataset_name} COMPLETO | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas"
        )

        print(
            f"Guardado: {output_file}"
        )


except KeyboardInterrupt:

    print(
        "\nPROCESO INTERRUMPIDO MANUALMENTE"
    )


except Exception as e:

    print(
        f"\nERROR: "
        f"{type(e).__name__}: {e}"
    )


finally:

    # ============================================================
    # ESTADO FINAL
    # ============================================================

    print("\n" + "=" * 70)
    print("ESTADO FINAL")
    print("=" * 70)


    if current_dataset is not None:

        print(
            f"Último dataset: {current_dataset}"
        )

        print(
            f"Última etapa: {current_stage}"
        )

        print(
            f"Chunks confirmados en la etapa: "
            f"{current_chunk}/{total_chunks}"
        )


    if completed_datasets:

        print(
            "\nDatasets completados:",
            ", ".join(
                completed_datasets
            )
        )

    else:

        print(
            "\nDatasets completados: ninguno"
        )


    if saved_files:

        print(
            "\nArchivos guardados exitosamente:"
        )

        for file in saved_files:

            print(
                f"  ✓ {file}"
            )

    else:

        print(
            "\nArchivos completos guardados: ninguno"
        )
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c1


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from __future__ import annotations

from dataclasses import dataclass
from typing import Sequence
import warnings

import numpy as np
import pandas as pd


# ============================================================
# DATASET DE TRABAJO
# ============================================================

DATASET = df_c3


# ============================================================
# CONFIGURACIÓN
# ============================================================

@dataclass(frozen=True)
class FeatureConfig:
    time_col: str = "bar_end"
    ys_col: str = "YS"

    price_suffix: str = "USDT"
    volume_suffix: str = "_v"
    trades_suffix: str = "_activity_trades_proxy"
    range_suffix: str = "_range_rel_proxy"
    illiq_suffix: str = "_illiquidity_proxy"
    active_suffix: str = "_active_trading"
    observed_suffix: str = "_observed"
    buy_suffix: str = "_taker_buy_usdt_proxy"
    sell_suffix: str = "_taker_sell_usdt_proxy"
    yp_suffix: str = "_yp"

    primary_window: int = 288
    min_periods: int = 72
    ewm_span: int = 72

    multiscale_windows: tuple[int, ...] = (
        3,
        12,
        72,
        288,
    )

    return_lag: int = 1

    stress_quantile: float = 0.90

    recovery_tolerance: float = 0.15
    recovery_hold_bars: int = 3
    recovery_event_ewm_alpha: float = 0.20

    eps: float = 1e-12
    dtype: str = "float32"

    asset_block_size: int = 32


cfg = FeatureConfig()


# ============================================================
# TIEMPO
# ============================================================

def validate_time(df, cfg):
    df = df.copy()

    df[cfg.time_col] = pd.to_datetime(
        df[cfg.time_col],
        utc=True,
        errors="raise"
    )

    df = (
        df.sort_values(cfg.time_col)
          .reset_index(drop=True)
    )

    if df[cfg.time_col].duplicated().any():
        raise ValueError("Existen timestamps duplicados.")

    delta = df[cfg.time_col].diff().dropna()

    expected = pd.Timedelta(minutes=5)

    if not delta.eq(expected).all():
        raise ValueError(
            "La serie no posee una grilla estricta de 5 minutos."
        )

    return df


# ============================================================
# ACTIVOS
# ============================================================

def discover_assets(df, cfg):

    price_cols = [
        c for c in df.columns
        if c.endswith(cfg.price_suffix)
        and "_" not in c
    ]

    candidates = sorted(
        c[:-len(cfg.price_suffix)]
        for c in price_cols
    )

    required_suffixes = (
        cfg.volume_suffix,
        cfg.trades_suffix,
        cfg.range_suffix,
        cfg.illiq_suffix,
        cfg.active_suffix,
        cfg.observed_suffix,
        cfg.buy_suffix,
        cfg.sell_suffix,
        cfg.yp_suffix,
    )

    assets = []

    for asset in candidates:

        if all(
            f"{asset}{suffix}" in df.columns
            for suffix in required_suffixes
        ):
            assets.append(asset)

    if len(assets) != 510:
        warnings.warn(
            f"Se detectaron {len(assets)} activos completos."
        )

    return assets


def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# NOMBRES
# ============================================================

def base_col(asset, kind, cfg):

    if kind == "P":
        return f"{asset}{cfg.price_suffix}"

    suffix = {
        "V": cfg.volume_suffix,
        "C": cfg.trades_suffix,
        "R": cfg.range_suffix,
        "L": cfg.illiq_suffix,
        "A": cfg.active_suffix,
        "O": cfg.observed_suffix,
        "B": cfg.buy_suffix,
        "S": cfg.sell_suffix,
        "yp": cfg.yp_suffix,
    }[kind]

    return f"{asset}{suffix}"


def feat_col(asset, feature):
    return f"{asset}__{feature}"


# ============================================================
# CARGA POR BLOQUES
# ============================================================

def load_base(df, assets, kind, cfg):

    cols = [
        base_col(a, kind, cfg)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def load_feature(df, assets, feature):

    cols = [
        feat_col(a, feature)
        for a in assets
    ]

    out = df.loc[:, cols].astype(
        "float64",
        copy=False
    )

    out.columns = assets

    return out


def name_block(x, feature, cfg):

    out = x.astype(
        cfg.dtype,
        copy=False
    )

    out.columns = [
        feat_col(a, feature)
        for a in out.columns
    ]

    return out


def merge_blocks(df, blocks):

    block = pd.concat(
        blocks,
        axis=1,
        copy=False
    )

    overlap = df.columns.intersection(
        block.columns
    )

    if len(overlap):
        df = df.drop(columns=list(overlap))

    return pd.concat(
        [df, block],
        axis=1,
        copy=False
    )


# ============================================================
# ROBUST Z CAUSAL
# ============================================================

def robust_z_causal(
    x,
    window,
    min_periods,
    eps
):

    hist = x.shift(1)

    roll = hist.rolling(
        window=window,
        min_periods=min_periods
    )

    median = roll.median()

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    scale = (
        q75 - q25
    ) / 1.349

    return (
        x - median
    ) / (
        scale + eps
    )


# ============================================================
# DISPERSIÓN ROBUSTA
# ============================================================

def rolling_robust_dispersion(
    x,
    window,
    min_periods,
    eps
):

    roll = x.rolling(
        window=window,
        min_periods=min_periods
    )

    q25 = roll.quantile(0.25)
    q75 = roll.quantile(0.75)

    return (
        (q75 - q25) / 1.349
        + eps
    )


# ============================================================
# AUTOCORRELACIÓN ROLLING
# ============================================================

def rolling_autocorr_lag1(
    x,
    window,
    min_periods
):

    lag = x.shift(1)

    result = {}

    for col in x.columns:
        result[col] = (
            x[col]
            .rolling(
                window=window,
                min_periods=min_periods
            )
            .corr(lag[col])
        )

    return pd.DataFrame(
        result,
        index=x.index
    )


# ============================================================
# BETA ROLLING
# ============================================================

def rolling_beta_with_intercept(
    x,
    y,
    window,
    min_periods,
    eps
):

    rx = x.rolling(
        window,
        min_periods=min_periods
    )

    ry = y.rolling(
        window,
        min_periods=min_periods
    )

    ex = rx.mean()
    ey = ry.mean()

    exy = (
        (x * y)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    ex2 = (
        (x * x)
        .rolling(
            window,
            min_periods=min_periods
        )
        .mean()
    )

    cov_xy = exy - ex * ey
    var_x = ex2 - ex * ex

    beta = (
        cov_xy
        / (var_x + eps)
    )

    return beta.clip(lower=0.0)


# ============================================================
# STRESS
# ============================================================

def rolling_stress_threshold(
    stress,
    window,
    min_periods,
    q
):

    return (
        stress
        .shift(1)
        .rolling(
            window,
            min_periods=min_periods
        )
        .quantile(q)
    )


def conditional_rolling_median_ratio(
    value,
    stress,
    threshold,
    window,
    min_periods,
    eps
):

    is_stress = (
        stress > threshold
    )

    numerator = (
        value
        .where(is_stress)
        .rolling(
            window,
            min_periods=max(
                5,
                min_periods // 10
            )
        )
        .median()
    )

    denominator = (
        value
        .rolling(
            window,
            min_periods=min_periods
        )
        .median()
    )

    return (
        numerator
        / (denominator + eps)
    )


# ============================================================
# PROXY POSITIVO LATENTE
# ============================================================

def softplus_df(x):

    values = np.logaddexp(
        0.0,
        x.to_numpy(
            dtype=np.float64,
            copy=False
        )
    )

    return pd.DataFrame(
        values,
        index=x.index,
        columns=x.columns
    )


def composite_positive_proxy(
    components,
    cfg
):

    total = None
    count = None

    for component in components:

        z = robust_z_causal(
            component,
            cfg.primary_window,
            cfg.min_periods,
            cfg.eps
        )

        valid = z.notna()

        if total is None:

            total = z.fillna(0.0)

            count = valid.astype(
                "float64"
            )

        else:

            total += z.fillna(0.0)

            count += valid.astype(
                "float64"
            )

    score = (
        total
        / count.replace(0.0, np.nan)
    )

    return softplus_df(score)


# ============================================================
# RECOVERY SPEED CAUSAL
# ============================================================

@njit(parallel=True)
def _recovery_speed_numba(
    g,
    s,
    q,
    b,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    T, N = g.shape
    output = np.full((T, N), np.nan, dtype=np.float64)

    for j in prange(N):
        in_event = False
        event_start = -1
        event_baseline = np.nan
        consecutive = 0
        duration_ewma = np.nan

        for t in range(T):
            if not in_event:
                if (
                    np.isfinite(s[t, j])
                    and np.isfinite(q[t, j])
                    and np.isfinite(b[t, j])
                    and s[t, j] > q[t, j]
                    and abs(b[t, j]) > eps
                ):
                    in_event = True
                    event_start = t
                    event_baseline = b[t, j]
                    consecutive = 0

            elif np.isfinite(g[t, j]):
                distance = abs(g[t, j] - event_baseline) / (
                    abs(event_baseline) + eps
                )

                if distance <= tolerance:
                    consecutive += 1
                else:
                    consecutive = 0

                if consecutive >= hold_bars:
                    recovery_start = t - hold_bars + 1
                    duration = max(1, recovery_start - event_start)

                    if np.isnan(duration_ewma):
                        duration_ewma = float(duration)
                    else:
                        duration_ewma = (
                            alpha * duration
                            + (1.0 - alpha) * duration_ewma
                        )

                    in_event = False
                    event_start = -1
                    event_baseline = np.nan
                    consecutive = 0

            if np.isfinite(duration_ewma):
                output[t, j] = 1.0 / (1.0 + duration_ewma)

    return output


def recovery_speed_causal(
    gp0,
    stress,
    stress_threshold,
    baseline,
    tolerance,
    hold_bars,
    alpha,
    eps
):
    frames = (stress, stress_threshold, baseline)

    if any(
        not gp0.index.equals(x.index)
        or not gp0.columns.equals(x.columns)
        for x in frames
    ):
        raise ValueError(
            "gp0, stress, stress_threshold y baseline deben compartir índice y columnas."
        )

    output = _recovery_speed_numba(
        gp0.to_numpy(dtype=np.float64, copy=False),
        stress.to_numpy(dtype=np.float64, copy=False),
        stress_threshold.to_numpy(dtype=np.float64, copy=False),
        baseline.to_numpy(dtype=np.float64, copy=False),
        tolerance,
        hold_bars,
        alpha,
        eps,
    )

    return pd.DataFrame(
        output,
        index=gp0.index,
        columns=gp0.columns,
    )

# ============================================================
# INICIALIZACIÓN
# ============================================================

DATASET = validate_time(DATASET, cfg)

assets = discover_assets(
    DATASET,
    cfg
)

print("Dataset:", DATASET.shape)
print("Activos detectados:", len(assets))
```

```python
from pathlib import Path
import os
import gc

# ============================================================
# CHUNKING
# ============================================================

def chunks(seq, size):
    for start in range(0, len(seq), size):
        yield list(seq[start:start + size])


# ============================================================
# DATASETS DE TRABAJO
# ============================================================

DATASETS = {
    "df_c1": df_c1,
    "df_c2": df_c2,
    "df_c3": df_c3,
}


# ============================================================
# ARCHIVOS DE SALIDA
# ============================================================

ruta = Path(
    "/content/drive/MyDrive/Neural/NPP/Cripto/Crypto 5 minutos/"
)

OUTPUTS = {
    "df_c1": ruta / "112024-042025_clean_add_final_Eth.parquet",
    "df_c2": ruta / "042025-092025_clean_add_final_Eth.parquet",
    "df_c3": ruta / "092025-022026_clean_add_final_Eth.parquet",
}


# ============================================================
# CONTROL DE PROCESO
# ============================================================

saved_files = []
completed_datasets = []

current_dataset = None
current_stage = None
current_chunk = 0
total_chunks = 0


try:

    for dataset_name, DATASET in DATASETS.items():

        current_dataset = dataset_name
        current_stage = "E_theory features"

        asset_chunks = list(
            chunks(
                assets,
                cfg.asset_block_size
            )
        )

        total_chunks = len(asset_chunks)
        current_chunk = 0

        print("\n" + "=" * 70)
        print(
            f"{dataset_name} | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas | "
            f"{total_chunks} chunks"
        )
        print("=" * 70)


        # ============================================================
        # E_theory GLOBAL
        # ============================================================

        E_all = load_feature(
            DATASET,
            assets,
            "E_theory"
        ).astype("float32")


        A_all = load_base(
            DATASET,
            assets,
            "A",
            cfg
        ).astype("float32")


        E_active = E_all.where(
            A_all > 0
        )


        # --------------------------------------------
        # Estadísticos transversales
        # --------------------------------------------

        E_median = E_active.median(
            axis=1
        )

        E_rank = E_active.rank(
            axis=1,
            pct=True,
            method="average"
        )


        # ============================================================
        # CÁLCULO POR CHUNK
        # ============================================================

        for i, chunk in enumerate(
            asset_chunks,
            start=1
        ):

            E = E_all.loc[
                :,
                chunk
            ]


            # --------------------------------------------
            # Log energía
            # --------------------------------------------

            logE = np.log(
                E.clip(
                    lower=cfg.eps
                )
            )


            # --------------------------------------------
            # Energía relativa transversal
            # --------------------------------------------

            Erel = E.sub(
                E_median,
                axis=0
            )


            # --------------------------------------------
            # Primera diferencia
            # --------------------------------------------

            dE = E.diff()


            # --------------------------------------------
            # Segunda diferencia
            # --------------------------------------------

            ddE = dE.diff()


            # --------------------------------------------
            # Shock robusto causal
            # --------------------------------------------

            ESH = robust_z_causal(
                E,
                cfg.primary_window,
                cfg.min_periods,
                cfg.eps
            )


            # --------------------------------------------
            # Ranking transversal
            # --------------------------------------------

            ERANK = E_rank.loc[
                :,
                chunk
            ]


            # ========================================================
            # PREPARACIÓN DEL CHUNK
            # ========================================================

            chunk_blocks = [
                name_block(
                    logE,
                    "Eth_logE",
                    cfg
                ),

                name_block(
                    Erel,
                    "Eth_Erel",
                    cfg
                ),

                name_block(
                    dE,
                    "Eth_dE",
                    cfg
                ),

                name_block(
                    ddE,
                    "Eth_ddE",
                    cfg
                ),

                name_block(
                    ESH,
                    "Eth_ESH",
                    cfg
                ),

                name_block(
                    ERANK,
                    "Eth_ERANK",
                    cfg
                ),
            ]


            # ========================================================
            # COMMIT DEL CHUNK
            # Solo se ejecuta si TODO el chunk terminó correctamente
            # ========================================================

            DATASET = merge_blocks(
                DATASET,
                chunk_blocks
            )

            DATASETS[dataset_name] = DATASET
            globals()[dataset_name] = DATASET

            current_chunk = i

            print(
                f"{dataset_name} | "
                f"chunk {i:02d}/{total_chunks} OK | "
                f"{len(chunk):02d} activos | "
                f"{DATASET.shape[1]:,} columnas"
            )


            del (
                E,
                logE,
                Erel,
                dE,
                ddE,
                ESH,
                ERANK,
                chunk_blocks,
            )

            gc.collect()


        # ============================================================
        # LIMPIEZA DE OBJETOS GLOBALES DEL DATASET
        # ============================================================

        del (
            E_all,
            A_all,
            E_active,
            E_median,
            E_rank,
        )

        gc.collect()


        # ============================================================
        # GUARDADO DEL DATASET COMPLETO
        # ============================================================

        output_file = OUTPUTS[
            dataset_name
        ]

        temp_file = output_file.with_suffix(
            ".tmp.parquet"
        )

        DATASET.to_parquet(
            temp_file,
            index=False
        )

        os.replace(
            temp_file,
            output_file
        )

        saved_files.append(
            str(output_file)
        )

        completed_datasets.append(
            dataset_name
        )

        print(
            f"\n{dataset_name} COMPLETO | "
            f"{DATASET.shape[0]:,} filas × "
            f"{DATASET.shape[1]:,} columnas"
        )

        print(
            f"Guardado: {output_file}"
        )


except KeyboardInterrupt:

    print(
        "\nPROCESO INTERRUMPIDO MANUALMENTE"
    )


except Exception as e:

    print(
        f"\nERROR: "
        f"{type(e).__name__}: {e}"
    )


finally:

    # ============================================================
    # ESTADO FINAL
    # ============================================================

    print("\n" + "=" * 70)
    print("ESTADO FINAL")
    print("=" * 70)


    if current_dataset is not None:

        print(
            f"Último dataset: {current_dataset}"
        )

        print(
            f"Última etapa: {current_stage}"
        )

        print(
            f"Chunks confirmados en la etapa: "
            f"{current_chunk}/{total_chunks}"
        )


    if completed_datasets:

        print(
            "\nDatasets completados:",
            ", ".join(
                completed_datasets
            )
        )

    else:

        print(
            "\nDatasets completados: ninguno"
        )


    if saved_files:

        print(
            "\nArchivos guardados exitosamente:"
        )

        for file in saved_files:

            print(
                f"  ✓ {file}"
            )

    else:

        print(
            "\nArchivos completos guardados: ninguno"
        )
```

```python
def chronological_splits(
    df,
    horizon=1,
    train_frac=0.70,
    val_frac=0.15,
    purge=None,
):
    if purge is None:
        purge = horizon

    n = len(df)

    # Últimas h filas no tienen target conocido
    usable_end = n - horizon

    train_end = int(
        usable_end * train_frac
    )

    val_end = int(
        usable_end
        * (train_frac + val_frac)
    )

    train_idx = np.arange(
        0,
        max(
            0,
            train_end - horizon
        )
    )

    val_idx = np.arange(
        train_end + purge,
        max(
            train_end + purge,
            val_end - horizon
        )
    )

    test_idx = np.arange(
        val_end + purge,
        usable_end
    )

    return {
        "train": train_idx,
        "val": val_idx,
        "test": test_idx,
    }


SPLITS = chronological_splits(
    df,
    horizon=1,
    purge=1,
)


for name, idx in SPLITS.items():

    print(
        name,
        len(idx),
        df[cfg.time_col].iloc[idx[0]],
        "→",
        df[cfg.time_col].iloc[idx[-1]],
    )
```

```python
def audit_node_feature(
    df,
    feature,
    assets,
    cfg,
    sample_rows=5000,
    seed=42,
):

    rng = np.random.default_rng(seed)

    if len(df) > sample_rows:
        row_sample = np.sort(
            rng.choice(
                len(df),
                sample_rows,
                replace=False,
            )
        )
    else:
        row_sample = np.arange(len(df))

    total = 0
    nan_count = 0
    inf_count = 0
    zero_count = 0
    support_bad = 0

    constant_assets = 0
    asset_count = 0

    sample_values = []

    lower, upper = SUPPORT.get(
        feature,
        (None, None),
    )

    for chunk in chunks(
        assets,
        cfg.asset_block_size,
    ):

        X = load_feature(
            df,
            chunk,
            feature,
        )

        arr = X.to_numpy(
            dtype=np.float64,
            copy=False,
        )

        total += arr.size

        nan_count += np.isnan(
            arr
        ).sum()

        inf_count += np.isinf(
            arr
        ).sum()

        finite = np.isfinite(arr)

        zero_count += (
            (arr == 0)
            & finite
        ).sum()

        if lower is not None:
            support_bad += (
                (arr < lower - 1e-7)
                & finite
            ).sum()

        if upper is not None:
            support_bad += (
                (arr > upper + 1e-7)
                & finite
            ).sum()

        # Activos degenerados
        for j in range(arr.shape[1]):

            x = arr[:, j]

            x = x[
                np.isfinite(x)
            ]

            asset_count += 1

            if (
                len(x) == 0
                or np.nanstd(x) < 1e-12
            ):
                constant_assets += 1

        sampled = arr[
            row_sample,
            :
        ].ravel()

        sampled = sampled[
            np.isfinite(sampled)
        ]

        if len(sampled):
            sample_values.append(
                sampled
            )

    if sample_values:
        sample_values = np.concatenate(
            sample_values
        )

        q = np.quantile(
            sample_values,
            [
                0.001,
                0.01,
                0.50,
                0.99,
                0.999,
            ]
        )

        q001, q01, med, q99, q999 = q

    else:
        q001 = q01 = med = q99 = q999 = np.nan

    return {
        "feature": feature,
        "family": NODE_FEATURES[feature],

        "nan_pct":
            100 * nan_count / total,

        "inf_pct":
            100 * inf_count / total,

        "zero_pct":
            100 * zero_count / total,

        "support_violation_pct":
            100 * support_bad / total,

        "constant_asset_pct":
            100 * constant_assets
            / max(asset_count, 1),

        "q_0.001": q001,
        "q_0.01": q01,
        "median": med,
        "q_0.99": q99,
        "q_0.999": q999,
    }


quality_rows = []

for feature in NODE_FEATURES:

    print("Auditando:", feature)

    quality_rows.append(
        audit_node_feature(
            df,
            feature,
            assets,
            cfg,
        )
    )


QUALITY_TABLE = (
    pd.DataFrame(quality_rows)
    .sort_values(
        [
            "family",
            "feature",
        ]
    )
    .reset_index(drop=True)
)


display(QUALITY_TABLE)
```

```python
def audit_system_features(
    df,
    features,
):

    rows = []

    for feature in features:

        x = (
            df[feature]
            .to_numpy(
                dtype=np.float64,
                copy=False
            )
        )

        finite = np.isfinite(x)

        xf = x[finite]

        rows.append({

            "feature": feature,

            "nan_pct":
                100
                * np.isnan(x).mean(),

            "inf_pct":
                100
                * np.isinf(x).mean(),

            "zero_pct":
                100
                * (
                    (x == 0)
                    & finite
                ).mean(),

            "median":
                np.nanmedian(xf)
                if len(xf)
                else np.nan,

            "IQR":
                (
                    np.nanquantile(
                        xf,
                        0.75
                    )
                    -
                    np.nanquantile(
                        xf,
                        0.25
                    )
                )
                if len(xf)
                else np.nan,

            "std":
                np.nanstd(xf)
                if len(xf)
                else np.nan,
        })

    return pd.DataFrame(rows)


SYSTEM_QUALITY = audit_system_features(
    df,
    SYSTEM_FEATURES,
)

display(SYSTEM_QUALITY)
```

```python
def sampled_node_panel(
    df,
    assets,
    features,
    cfg,
    n_times=3000,
    n_assets=96,
    seed=42,
):

    rng = np.random.default_rng(seed)

    chosen_assets = (
        list(assets)
        if len(assets) <= n_assets
        else list(
            rng.choice(
                assets,
                n_assets,
                replace=False,
            )
        )
    )

    valid_rows = np.arange(
        0,
        len(df) - 1
    )

    if len(valid_rows) > n_times:

        time_idx = np.sort(
            rng.choice(
                valid_rows,
                n_times,
                replace=False,
            )
        )

    else:
        time_idx = valid_rows

    data = {}

    for feature in features:

        X = load_feature(
            df,
            chosen_assets,
            feature,
        )

        data[feature] = (
            X.iloc[time_idx]
            .to_numpy(
                dtype=np.float32,
                copy=False
            )
            .reshape(-1)
        )

    yp_cols = [
        f"{a}{cfg.yp_suffix}"
        for a in chosen_assets
    ]

    yp = (
        df.loc[
            df.index[time_idx],
            yp_cols,
        ]
        .to_numpy(
            dtype=np.float32,
            copy=False
        )
        .reshape(-1)
    )

    panel = pd.DataFrame(data)

    panel["yp_t"] = yp

    panel["log_yp_t"] = np.log(
        np.clip(
            yp,
            cfg.eps,
            None,
        )
    )

    return panel


AUDIT_PANEL = sampled_node_panel(
    df,
    assets,
    STRICT_FEATURES,
    cfg,
)
```

```python
def redundancy_report(
    panel,
    features,
    threshold=0.97,
):

    X = panel[features]

    corr = X.corr(
        method="spearman"
    )

    rows = []

    for i in range(len(features)):

        for j in range(
            i + 1,
            len(features)
        ):

            a = features[i]
            b = features[j]

            rho = corr.loc[a, b]

            if (
                np.isfinite(rho)
                and abs(rho) >= threshold
            ):

                rows.append({
                    "feature_1": a,
                    "feature_2": b,
                    "spearman": rho,
                })

    return (
        pd.DataFrame(rows)
        .sort_values(
            "spearman",
            key=np.abs,
            ascending=False,
        )
        if rows
        else pd.DataFrame(
            columns=[
                "feature_1",
                "feature_2",
                "spearman",
            ]
        )
    )


REDUNDANCY = redundancy_report(
    AUDIT_PANEL,
    STRICT_FEATURES,
    threshold=0.97,
)

display(REDUNDANCY)
```

```python
def circularity_report(
    panel,
    features,
    max_samples=50000,
    seed=42,
):

    rng = np.random.default_rng(seed)

    work = panel[
        features
        + ["log_yp_t"]
    ].replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    if len(work) > max_samples:

        idx = rng.choice(
            len(work),
            max_samples,
            replace=False,
        )

        work = work.iloc[idx]

    y = work[
        "log_yp_t"
    ].to_numpy()

    rows = []

    for feature in features:

        x = work[
            feature
        ].to_numpy()

        # Spearman
        rho = (
            pd.Series(x)
            .corr(
                pd.Series(y),
                method="spearman",
            )
        )

        # Información mutua
        mi = mutual_info_regression(
            x.reshape(-1, 1),
            y,
            random_state=seed,
        )[0]

        # Capacidad lineal de reconstrucción
        split = int(
            len(x) * 0.70
        )

        model = LinearRegression()

        model.fit(
            x[:split].reshape(-1, 1),
            y[:split],
        )

        pred = model.predict(
            x[split:].reshape(-1, 1)
        )

        r2 = r2_score(
            y[split:],
            pred,
        )

        rows.append({
            "feature": feature,
            "family": NODE_FEATURES[feature],
            "spearman_logyp": rho,
            "MI_logyp": mi,
            "linear_R2_logyp": r2,
        })

    return (
        pd.DataFrame(rows)
        .sort_values(
            "MI_logyp",
            ascending=False,
        )
        .reset_index(drop=True)
    )


CIRCULARITY = circularity_report(
    AUDIT_PANEL,
    STRICT_FEATURES,
)

display(CIRCULARITY)
```

```python
def temporal_stability(
    df,
    assets,
    features,
    cfg,
    n_blocks=4,
    n_assets=64,
    rows_per_block=3000,
    seed=42,
):

    rng = np.random.default_rng(seed)

    chosen_assets = (
        list(assets)
        if len(assets) <= n_assets
        else list(
            rng.choice(
                assets,
                n_assets,
                replace=False,
            )
        )
    )

    blocks_idx = np.array_split(
        np.arange(len(df)),
        n_blocks,
    )

    output = []

    for feature in features:

        distributions = []

        for b, idx in enumerate(
            blocks_idx
        ):

            if len(idx) > rows_per_block:

                sample_idx = np.sort(
                    rng.choice(
                        idx,
                        rows_per_block,
                        replace=False,
                    )
                )

            else:
                sample_idx = idx

            X = load_feature(
                df,
                chosen_assets,
                feature,
            )

            values = (
                X.iloc[sample_idx]
                .to_numpy(
                    dtype=np.float32,
                    copy=False,
                )
                .reshape(-1)
            )

            values = values[
                np.isfinite(values)
            ]

            distributions.append(
                values
            )

        base = distributions[0]

        base_iqr = (
            np.quantile(base, 0.75)
            - np.quantile(base, 0.25)
        )

        for b, values in enumerate(
            distributions
        ):

            med = np.median(values)

            iqr = (
                np.quantile(
                    values,
                    0.75
                )
                -
                np.quantile(
                    values,
                    0.25
                )
            )

            if b == 0:
                w_norm = 0.0
            else:
                w = wasserstein_distance(
                    base,
                    values,
                )

                w_norm = (
                    w
                    /
                    (
                        base_iqr
                        + cfg.eps
                    )
                )

            output.append({
                "feature": feature,
                "block": b,
                "median": med,
                "IQR": iqr,
                "wasserstein_vs_block0_IQR": w_norm,
            })

    return pd.DataFrame(output)


STABILITY = temporal_stability(
    df,
    assets,
    STRICT_FEATURES,
    cfg,
)

display(STABILITY)
```

```python
from statsmodels.tsa.stattools import (
    adfuller,
    kpss,
)


def stationarity_screen(
    df,
    assets,
    features,
    cfg,
):

    rows = []

    for feature in features:

        X = load_feature(
            df,
            assets,
            feature,
        )

        x = (
            X.median(
                axis=1,
                skipna=True,
            )
            .replace(
                [np.inf, -np.inf],
                np.nan,
            )
            .dropna()
        )

        if len(x) < 100:
            continue

        try:
            adf_p = adfuller(
                x,
                autolag="AIC",
            )[1]
        except Exception:
            adf_p = np.nan

        try:
            kpss_p = kpss(
                x,
                regression="c",
                nlags="auto",
            )[1]
        except Exception:
            kpss_p = np.nan

        rows.append({
            "feature": feature,
            "ADF_p": adf_p,
            "KPSS_p": kpss_p,
        })

    return pd.DataFrame(rows)


STATIONARITY = stationarity_screen(
    df,
    assets,
    STRICT_FEATURES,
    cfg,
)

display(STATIONARITY)
```

```python
def yp_matrix(
    df,
    assets,
    rows,
    cfg,
    horizon=0,
):

    rows = np.asarray(rows)

    target_rows = (
        rows + horizon
    )

    cols = [
        f"{a}{cfg.yp_suffix}"
        for a in assets
    ]

    return (
        df.iloc[target_rows][cols]
        .to_numpy(
            dtype=np.float64,
            copy=False,
        )
    )


def normalize_simplex(
    X,
    eps=1e-12,
    smooth=False,
):

    X = np.nan_to_num(
        X,
        nan=0.0,
        posinf=0.0,
        neginf=0.0,
    )

    X = np.clip(
        X,
        0.0,
        None,
    )

    if smooth:
        X = X + eps

    denom = X.sum(
        axis=1,
        keepdims=True,
    )

    return (
        X
        /
        np.maximum(
            denom,
            eps,
        )
    )


def simplex_metrics(
    y,
    q,
    eps=1e-12,
):

    y = normalize_simplex(
        y,
        eps,
        smooth=False,
    )

    q = normalize_simplex(
        q,
        eps,
        smooth=True,
    )

    log_y = np.log(
        np.clip(
            y,
            eps,
            None,
        )
    )

    log_q = np.log(
        np.clip(
            q,
            eps,
            None,
        )
    )

    CE_t = -np.sum(
        y * log_q,
        axis=1,
    )

    KL_t = np.sum(
        y * (
            log_y
            - log_q
        ),
        axis=1,
    )

    m = 0.5 * (
        y + q
    )

    log_m = np.log(
        np.clip(
            m,
            eps,
            None,
        )
    )

    JS_t = 0.5 * (
        np.sum(
            y * (
                log_y - log_m
            ),
            axis=1,
        )
        +
        np.sum(
            q * (
                log_q - log_m
            ),
            axis=1,
        )
    )

    return {
        "CE": CE_t.mean(),
        "KL": KL_t.mean(),
        "JS": JS_t.mean(),
        "MAE": np.mean(
            np.abs(
                y - q
            )
        ),
        "RMSE": np.sqrt(
            np.mean(
                (y - q) ** 2
            )
        ),
        "CE_t": CE_t,
    }
```

```python
def sample_rows_evenly(
    idx,
    max_rows,
):

    idx = np.asarray(idx)

    if len(idx) <= max_rows:
        return idx

    positions = np.linspace(
        0,
        len(idx) - 1,
        max_rows,
        dtype=int,
    )

    return idx[positions]


def load_feature_matrix(
    df,
    assets,
    feature,
    rows,
):

    cols = [
        f"{a}__{feature}"
        for a in assets
    ]

    return (
        df.iloc[rows][cols]
        .to_numpy(
            dtype=np.float64,
            copy=False,
        )
    )


def fit_robust_scaler(
    X,
    eps=1e-12,
):

    values = X[
        np.isfinite(X)
    ]

    median = np.median(
        values
    )

    q25, q75 = np.quantile(
        values,
        [0.25, 0.75]
    )

    scale = (
        q75 - q25
    ) / 1.349

    return (
        median,
        max(scale, eps),
    )


def transform_feature(
    X,
    median,
    scale,
):

    Z = (
        X - median
    ) / scale

    Z = np.nan_to_num(
        Z,
        nan=0.0,
        posinf=8.0,
        neginf=-8.0,
    )

    return np.clip(
        Z,
        -8.0,
        8.0,
    )


def tilted_simplex(
    p,
    x,
    beta,
    eps=1e-12,
):

    p = normalize_simplex(
        p,
        eps,
        smooth=True,
    )

    logits = (
        np.log(
            np.clip(
                p,
                eps,
                None,
            )
        )
        +
        beta * x
    )

    return softmax(
        logits,
        axis=1,
    )
```

```python
def screen_single_feature(
    df,
    assets,
    feature,
    cfg,
    splits,
    horizon=1,
    train_max_rows=15000,
):

    # --------------------------------------------
    # Train para beta
    # --------------------------------------------

    train_rows = sample_rows_evenly(
        splits["train"],
        train_max_rows,
    )

    # Evitar desbordar final
    train_rows = train_rows[
        train_rows + horizon
        < len(df)
    ]

    X_train = load_feature_matrix(
        df,
        assets,
        feature,
        train_rows,
    )

    med, scale = fit_robust_scaler(
        X_train,
        cfg.eps,
    )

    X_train = transform_feature(
        X_train,
        med,
        scale,
    )

    p_train = yp_matrix(
        df,
        assets,
        train_rows,
        cfg,
        horizon=0,
    )

    y_train = yp_matrix(
        df,
        assets,
        train_rows,
        cfg,
        horizon=horizon,
    )

    p_train = normalize_simplex(
        p_train,
        cfg.eps,
        smooth=True,
    )

    y_train = normalize_simplex(
        y_train,
        cfg.eps,
        smooth=False,
    )


    def objective(beta):

        q = tilted_simplex(
            p_train,
            X_train,
            beta,
            cfg.eps,
        )

        return simplex_metrics(
            y_train,
            q,
            cfg.eps,
        )["CE"]


    result = minimize_scalar(
        objective,
        bounds=(-5.0, 5.0),
        method="bounded",
    )

    beta = result.x


    # --------------------------------------------
    # Test completo
    # --------------------------------------------

    test_rows = splits["test"]

    test_rows = test_rows[
        test_rows + horizon
        < len(df)
    ]

    X_test = load_feature_matrix(
        df,
        assets,
        feature,
        test_rows,
    )

    X_test = transform_feature(
        X_test,
        med,
        scale,
    )

    p_test = yp_matrix(
        df,
        assets,
        test_rows,
        cfg,
        horizon=0,
    )

    y_test = yp_matrix(
        df,
        assets,
        test_rows,
        cfg,
        horizon=horizon,
    )

    p_test = normalize_simplex(
        p_test,
        cfg.eps,
        smooth=True,
    )

    y_test = normalize_simplex(
        y_test,
        cfg.eps,
        smooth=False,
    )

    # Persistence
    baseline_metrics = simplex_metrics(
        y_test,
        p_test,
        cfg.eps,
    )

    # Feature
    q_test = tilted_simplex(
        p_test,
        X_test,
        beta,
        cfg.eps,
    )

    feature_metrics = simplex_metrics(
        y_test,
        q_test,
        cfg.eps,
    )

    return {

        "feature": feature,
        "family": NODE_FEATURES[feature],
        "horizon_bars": horizon,
        "horizon_minutes": horizon * 5,

        "beta": beta,

        "CE_baseline":
            baseline_metrics["CE"],

        "CE_feature":
            feature_metrics["CE"],

        "delta_CE":
            baseline_metrics["CE"]
            - feature_metrics["CE"],

        "KL":
            feature_metrics["KL"],

        "JS":
            feature_metrics["JS"],

        "MAE":
            feature_metrics["MAE"],

        "RMSE":
            feature_metrics["RMSE"],

        # Necesario para bootstrap
        "delta_CE_t":
            baseline_metrics["CE_t"]
            - feature_metrics["CE_t"],
    }
```

```python
def moving_block_bootstrap(
    x,
    block_size=288,
    n_boot=1000,
    seed=42,
):

    rng = np.random.default_rng(
        seed
    )

    x = np.asarray(
        x,
        dtype=float,
    )

    x = x[
        np.isfinite(x)
    ]

    n = len(x)

    if n < block_size:
        raise ValueError(
            "Serie demasiado corta."
        )

    means = np.empty(
        n_boot
    )

    max_start = (
        n - block_size
    )

    n_blocks = int(
        np.ceil(
            n / block_size
        )
    )

    for b in range(n_boot):

        starts = rng.integers(
            0,
            max_start + 1,
            size=n_blocks,
        )

        sample = np.concatenate([
            x[
                s:s + block_size
            ]
            for s in starts
        ])[:n]

        means[b] = sample.mean()

    return {
        "mean": x.mean(),

        "CI_2.5":
            np.quantile(
                means,
                0.025,
            ),

        "CI_97.5":
            np.quantile(
                means,
                0.975,
            ),

        "P_mean_gt_0":
            np.mean(
                means > 0
            ),
    }
```

```python
def build_family_panel(
    df,
    assets,
    rows,
    node_features,
    system_features=None,
    cfg=None,
    horizon=1,
):

    rows = np.asarray(rows)

    rows = rows[
        rows + horizon
        < len(df)
    ]

    N = len(assets)
    T = len(rows)

    X_parts = {}

    # --------------------------------------------
    # Features nodales
    # --------------------------------------------

    for feature in node_features:

        M = load_feature_matrix(
            df,
            assets,
            feature,
            rows,
        )

        X_parts[feature] = (
            M.reshape(-1)
        )

    # --------------------------------------------
    # yp actual / futuro
    # --------------------------------------------

    p = yp_matrix(
        df,
        assets,
        rows,
        cfg,
        horizon=0,
    )

    y = yp_matrix(
        df,
        assets,
        rows,
        cfg,
        horizon=horizon,
    )

    p = normalize_simplex(
        p,
        cfg.eps,
        smooth=True,
    )

    y = normalize_simplex(
        y,
        cfg.eps,
        smooth=True,
    )

    log_p = np.log(
        np.clip(
            p,
            cfg.eps,
            None,
        )
    )

    delta_target = (
        np.log(
            np.clip(
                y,
                cfg.eps,
                None,
            )
        )
        - log_p
    )

    # Baseline importante
    X_parts["log_yp_t"] = (
        log_p.reshape(-1)
    )

    # --------------------------------------------
    # Contexto sistémico
    # --------------------------------------------

    if system_features:

        for feature in system_features:

            values = (
                df[feature]
                .iloc[rows]
                .to_numpy(
                    dtype=np.float32,
                    copy=False,
                )
            )

            X_parts[
                f"SYS__{feature}"
            ] = np.repeat(
                values,
                N,
            )

    X = pd.DataFrame(
        X_parts
    )

    target = (
        delta_target
        .reshape(-1)
    )

    # Información necesaria para reconstruir
    metadata = {
        "rows": rows,
        "n_times": T,
        "n_assets": N,
        "p": p,
        "y": y,
    }

    valid = (
        np.isfinite(target)
    )

    # Para árboles, dejamos NaN en features:
    # XGB/LGBM los pueden manejar.
    return (
        X.loc[valid].reset_index(drop=True),
        target[valid],
        metadata,
        valid,
    )
```

```python
def evaluate_family_delta(
    delta_prediction,
    metadata,
    valid,
    eps=1e-12,
):

    T = metadata["n_times"]
    N = metadata["n_assets"]

    full_delta = np.zeros(
        T * N,
        dtype=np.float64,
    )

    full_delta[valid] = (
        delta_prediction
    )

    delta = full_delta.reshape(
        T,
        N,
    )

    p = metadata["p"]
    y = metadata["y"]

    logits = (
        np.log(
            np.clip(
                p,
                eps,
                None,
            )
        )
        + delta
    )

    q = softmax(
        logits,
        axis=1,
    )

    baseline = simplex_metrics(
        y,
        p,
        eps,
    )

    model = simplex_metrics(
        y,
        q,
        eps,
    )

    return {
        "CE_baseline":
            baseline["CE"],

        "CE_model":
            model["CE"],

        "delta_CE":
            baseline["CE"]
            - model["CE"],

        "KL":
            model["KL"],

        "JS":
            model["JS"],

        "MAE":
            model["MAE"],

        "RMSE":
            model["RMSE"],
    }
```

```python
def sampled_split_rows(
    idx,
    n_times,
):

    return sample_rows_evenly(
        idx,
        n_times,
    )


XGB_ROWS = {
    "train": sampled_split_rows(
        SPLITS["train"],
        2500,
    ),

    "val": sampled_split_rows(
        SPLITS["val"],
        750,
    ),

    "test": sampled_split_rows(
        SPLITS["test"],
        1000,
    ),
}
```

```python
from xgboost import XGBRegressor


def run_xgb_family(
    family,
    df,
    assets,
    cfg,
    use_system_context=True,
):

    features = MODEL_FAMILIES[
        family
    ]

    system = (
        SYSTEM_FEATURES
        if use_system_context
        else None
    )

    X_train, y_train, _, _ = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["train"],
            features,
            system,
            cfg,
        )
    )

    X_val, y_val, _, _ = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["val"],
            features,
            system,
            cfg,
        )
    )

    X_test, y_test, meta_test, valid_test = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["test"],
            features,
            system,
            cfg,
        )
    )

    model = XGBRegressor(
        objective="reg:squarederror",

        n_estimators=1200,
        learning_rate=0.03,

        max_depth=6,
        min_child_weight=20,

        subsample=0.80,
        colsample_bytree=0.80,

        reg_alpha=0.05,
        reg_lambda=1.0,

        tree_method="hist",

        n_jobs=-1,

        random_state=42,
    )

    model.fit(
        X_train,
        y_train,

        eval_set=[
            (X_val, y_val)
        ],

        verbose=False,
    )

    pred = model.predict(
        X_test
    )

    metrics = evaluate_family_delta(
        pred,
        meta_test,
        valid_test,
        cfg.eps,
    )

    metrics.update({
        "family": family,
        "model": "XGBoost",
        "n_features": len(features),
    })

    return (
        model,
        metrics,
    )
```

```python
XGB_MODELS = {}
XGB_RESULTS = []


for family in MODEL_FAMILIES:

    print(
        "XGBoost:",
        family
    )

    model, metrics = run_xgb_family(
        family,
        df,
        assets,
        cfg,
        use_system_context=True,
    )

    XGB_MODELS[family] = model

    XGB_RESULTS.append(
        metrics
    )


XGB_RESULTS = pd.DataFrame(
    XGB_RESULTS
).sort_values(
    "delta_CE",
    ascending=False,
)


display(XGB_RESULTS)
```

```python
def run_lgb_family(
    family,
    df,
    assets,
    cfg,
    use_system_context=True,
):

    if not LIGHTGBM_AVAILABLE:
        raise ImportError(
            "Instala lightgbm antes de ejecutar."
        )

    features = MODEL_FAMILIES[
        family
    ]

    system = (
        SYSTEM_FEATURES
        if use_system_context
        else None
    )

    X_train, y_train, _, _ = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["train"],
            features,
            system,
            cfg,
        )
    )

    X_val, y_val, _, _ = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["val"],
            features,
            system,
            cfg,
        )
    )

    X_test, y_test, meta_test, valid_test = (
        build_family_panel(
            df,
            assets,
            XGB_ROWS["test"],
            features,
            system,
            cfg,
        )
    )

    model = LGBMRegressor(

        objective="regression",

        n_estimators=2000,
        learning_rate=0.025,

        num_leaves=63,
        max_depth=-1,

        min_child_samples=100,

        subsample=0.80,
        colsample_bytree=0.80,

        reg_alpha=0.05,
        reg_lambda=1.0,

        n_jobs=-1,

        random_state=42,
        verbosity=-1,
    )

    model.fit(

        X_train,
        y_train,

        eval_set=[
            (X_val, y_val)
        ],

        callbacks=[
            early_stopping(
                75,
                verbose=False,
            ),
            log_evaluation(0),
        ],
    )

    pred = model.predict(
        X_test
    )

    metrics = evaluate_family_delta(
        pred,
        meta_test,
        valid_test,
        cfg.eps,
    )

    metrics.update({
        "family": family,
        "model": "LightGBM",
        "n_features": len(features),
    })

    return (
        model,
        metrics,
    )
```

```python
def prefix_invariance_test(
    raw_df,
    rebuild_fn,
    feature_cols,
    cuts,
    atol=1e-7,
    rtol=1e-6,
):

    # Pipeline completo
    full = rebuild_fn(
        raw_df.copy()
    )

    results = []

    for cut in cuts:

        prefix_raw = (
            raw_df
            .iloc[:cut + 1]
            .copy()
        )

        prefix = rebuild_fn(
            prefix_raw
        )

        for feature in feature_cols:

            a = full[
                feature
            ].iloc[cut]

            b = prefix[
                feature
            ].iloc[-1]

            if (
                pd.isna(a)
                and pd.isna(b)
            ):
                equal = True
            else:
                equal = np.isclose(
                    a,
                    b,
                    atol=atol,
                    rtol=rtol,
                    equal_nan=True,
                )

            results.append({
                "cut": cut,
                "feature": feature,
                "full": a,
                "prefix": b,
                "causal_equal": equal,
            })

    return pd.DataFrame(
        results
    )
```
